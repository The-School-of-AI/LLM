### 6. Parse Metrics

```python
import json
import os

RESULT_DIR = "results_ifeval"
metrics_file = os.path.join(RESULT_DIR, "metrics.json")

with open(metrics_file, "r") as f:
    data = json.load(f)

print("\n===== IFEval Metrics =====\n")

# Extract first task
task = data["tasks"][0]

metrics = task["metrics"]
primary_score = metrics["primary_score"]

print(f"Model: {data['model_config']['model']}")
print(f"Primary IFEval Score: {primary_score:.4f}")

print("\n--- Key Metrics ---")
print(f"Instruction Loose Accuracy: {metrics['inst_level_loose_acc']:.4f}")
print(f"Instruction Strict Accuracy: {metrics['inst_level_strict_acc']:.4f}")
print(f"Prompt Loose Accuracy: {metrics['prompt_level_loose_acc']:.4f}")
print(f"Prompt Strict Accuracy: {metrics['prompt_level_strict_acc']:.4f}")

# sanity check
if primary_score < 0.1:
    print("\n⚠️ Very low score — likely model too weak or formatting issue.")
elif primary_score > 0.9:
    print("\n⚠️ Suspiciously high score — check evaluation.")
else:
    print("\n✔ Metric parsing successful.")
```

### 7. Inspect Per-Sample Predictions
Note: only first 3 prompts are saved. Output and per-sample scores available for all 541.

```python
import json
import random

PRED_FILE = "/content/results_ifeval/task-000-ifeval-predictions.jsonl"
INPUT_FILE = "/content/results_ifeval/task-000-ifeval-recorded-inputs.jsonl"

with open(PRED_FILE, "r") as f:
    predictions = [json.loads(line) for line in f]

# Only first 3 inputs are recorded (num_recorded_inputs=3)
with open(INPUT_FILE, "r") as f:
    inputs = {r["doc_id"]: r for r in (json.loads(line) for line in f)}

print(f"Total predictions: {len(predictions)}")
print(f"Recorded inputs available: {len(inputs)}")

for i, s in enumerate(random.sample(predictions, 10)):
    doc_id = s["doc_id"]
    print(f"
=== Sample {i+1} (doc_id={doc_id}) ===")
    if doc_id in inputs:
        prompt = inputs[doc_id].get("request", {}).get("prompt", "N/A")
        print(f"Prompt: {prompt[:200]}")
    else:
        print("Prompt: [not recorded]")
    output = s.get("model_output", [{}])[0].get("continuation", "N/A")
    print(f"Output: {output[:200]}")
    inst_loose = s.get("metrics", {}).get("inst_level_loose_acc", "N/A")
    print(f"inst_level_loose_acc: {inst_loose}")
```
